```python
def get_match_ids_by_puuid(puuid, region="europe", count=50):
    url = f"https://{region}.api.riotgames.com/lol/match/v5/matches/by-puuid/{puuid}/ids"
    params = {"start": 0, "count": count}
    r = requests.get(url, headers=HEADERS, params=params)
    r.raise_for_status()
    return r.json()
```

```python
def extract_player_stats(match_data, puuid):
    for p in match_data["info"]["participants"]:
        if p["puuid"] == puuid:
            return {
                "championId": p["championId"],
                "win": p["win"],
                "kills": p["kills"],
                "deaths": p["deaths"],
                "assists": p["assists"]
            }
    return None
```

```python
def get_champion_stats(puuid, match_count=50):
    match_ids = get_match_ids_by_puuid(puuid, count=match_count)
    stats = []

    for match_id in match_ids:
        match = get_match_details(match_id)
        player_stats = extract_player_stats(match, puuid)
        if player_stats:
            stats.append(player_stats)

    return stats
```

```python
import pandas as pd

def winrate_by_champion(puuid, match_count=50):
    stats = get_champion_stats(puuid, match_count)
    df = pd.DataFrame(stats)

    grouped = df.groupby("championId").agg(
        games=("win", "count"),
        wins=("win", "sum")
    )

    grouped["winrate"] = (grouped["wins"] / grouped["games"] * 100).round(2)
    return grouped.sort_values("games", ascending=False)
```

```python
def get_match_details(match_id, region="europe"):
    url = f"https://{region}.api.riotgames.com/lol/match/v5/matches/{match_id}"
    response = requests.get(url, headers=HEADERS)

    if response.status_code != 200:
        raise Exception(f"Erreur Riot API: {response.status_code} - {response.text}")

    return response.json()
```

```python
winrates = winrate_by_champion(puuid, match_count=100)
winrates
```
