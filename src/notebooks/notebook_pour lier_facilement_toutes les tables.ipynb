```python
import os
import glob
import pandas as pd
from collections import defaultdict

def print_csv_line_stats(folder_path):
    # dictionnaire pour stocker stats par sous-dossier
    stats = defaultdict(lambda: defaultdict(int))

    # récupérer tous les fichiers csv récursivement
    all_files = glob.glob(os.path.join(folder_path, "**/*.csv"), recursive=True)

    for f in all_files:
        # trouver le sous-dossier relatif
        subfolder = os.path.relpath(os.path.dirname(f), folder_path)
        try:
            df = pd.read_csv(f)
            n_lines = len(df)
            stats[subfolder]['total_files'] += 1
            stats[subfolder][f"{n_lines}_lines"] += 1
        except Exception as e:
            print(f"Erreur avec le fichier {f}: {e}")

    # print des stats
    for subfolder, counts in stats.items():
        print(f"--- {subfolder} ---")
        print(f"Total files: {counts['total_files']}")
        for k, v in counts.items():
            if k != 'total_files':
                print(f"Files with {k}: {v}")
        print()

    return stats

# Exemple d'utilisation
folder_path = os.path.join(PATH, "matchups/")
stats = print_csv_line_stats(folder_path)
```

```python
import re
import unicodedata

def normalize_champion_name(name):
    if pd.isna(name):
        return name
    
    # passer en string
    name = str(name)
    
    # enlever accents
    name = unicodedata.normalize("NFKD", name).encode("ascii", "ignore").decode("utf-8")
    
    # lower case
    name = name.lower()
    
    # enlever tout sauf lettres et chiffres
    name = re.sub(r"[^a-z0-9]", "", name)
    
    return name
```

```python
def normalize_champs_names(df_champs):

    df_champs = df_champs.copy()

    df_champs["name"] = df_champs["name"].apply(normalize_champion_name)

    return df_champs
```

```python
def normalize_games_champions(df_games, champion_cols):

    df_games = df_games.copy()

    for col in champion_cols:
        df_games[col] = df_games[col].apply(normalize_champion_name)

    return df_games
```

```python
# aggrege games et champions : 

import pandas as pd

# --- normaliser les patches ---
df_games = df_games.copy()
df_champion = df_champs.copy()

df_games["patch_short"] = df_games["patch"].str.extract(r"(\d+\.\d+)")
df_champion["patch_short"] = df_champion["patch"].str.extract(r"(\d+\.\d+)")

# colonnes champion dans games
champion_cols = [
    "blue_top_champion",
    "blue_jungle_champion",
    "blue_mid_champion",
    "blue_adc_champion",
    "blue_support_champion",
    "red_top_champion",
    "red_jungle_champion",
    "red_mid_champion",
    "red_adc_champion",
    "red_support_champion",
]

# stats champion (tout sauf patch / name)
champ_stats = df_champion.columns.difference(["patch", "patch_short", "name"])

df_result = df_games.copy()

for col in champion_cols:

    prefix = col.replace("_champion", "")

    temp = df_champion.rename(columns={"name": col})

    temp = temp[["patch_short", col] + list(champ_stats)]

    temp = temp.rename(
        columns={c: f"{prefix}_{c}" for c in champ_stats}
    )

    df_result = df_result.merge(
        temp,
        how="left",
        on=["patch_short", col],
    )

# optionnel : supprimer patch_short
df_result = df_result.drop(columns="patch_short")
```

```python
import pandas as pd

def reformat_matchups(matchups_df):
    result_rows = []

    # les lanes de matchups
    lanes = ["adc", "jun", "mid", "sup", "top"]

    for _, row in matchups_df.iterrows():
        champion = row["champion"]
        role = row["role"]

        for lane in lanes:
            # récupérer toutes les colonnes correspondant à cette lane
            lane_cols = [col for col in matchups_df.columns if col.startswith(f"matchup_{lane}_")]
            n_matchups = len(lane_cols) // 4  # chaque matchup a 4 colonnes

            for i in range(1, n_matchups + 1):
                name_col = f"matchup_{lane}_{i}_name"
                winrate_col = f"matchup_{lane}_{i}_winrate"
                games_col = f"matchup_{lane}_{i}_games"
                lane_quality_col = f"matchup_{lane}_{i}_lane_quality"

                if pd.notna(row.get(name_col)):
                    result_rows.append({
                        "champion": champion,
                        "role": role,
                        "role_matchup": lane,
                        "matchup_name": row.get(name_col),
                        "matchup_winrate": row.get(winrate_col),
                        "matchup_games": row.get(games_col),
                        "matchup_lane_quality": row.get(lane_quality_col),
                        "role_play_ratio": row["role_play_ratio"],
                        "tier": row["tier"],
                        "winrate": row["winrate"],
                        "pickrate": row["pickrate"],
                        "banrate": row["banrate"],
                        "nb_games_analyzed": row["nb_games_analyzed"]
                    })

    df_long = pd.DataFrame(result_rows)
    return df_long
```

```python
import pandas as pd

def build_champion_index_vector(df_simple_wr_norm):
    """
    Retourne un DataFrame de tous les champions concaténés
    avec un MultiIndex: champion
    """
    df_simple_wr_norm_copy = df_simple_wr_norm.copy()
    df_simple_wr_norm_copy["patch_float"] = df_simple_wr_norm_copy["patch"].astype(float)
    df_simple_wr_norm_copy.set_index("champion", inplace=True)
    return df_simple_wr_norm_copy
```

```python
def stack_champions(df_result_norm):
    champ_cols = [
        "blue_top_champion","blue_jun_champion","blue_mid_champion",
        "blue_adc_champion","blue_sup_champion",
        "red_top_champion","red_jun_champion","red_mid_champion",
        "red_adc_champion","red_sup_champion"
    ]

    df_stacked = df_result_norm.melt(
        id_vars=["match_id","server","patch","elo","blue_side_win"],
        value_vars=champ_cols,
        var_name="role",
        value_name="champion"
    )

    df_stacked["patch_float"] = df_stacked["patch"].astype(float)

    return df_stacked
```

```python
def score_candidates(df_stacked, df_simple_wr_norm_indexed):
    import numpy as np

    # Merge stacked avec simple_wr pour tous les champions
    df_merged = df_stacked.merge(
        df_simple_wr_norm_indexed.reset_index(),
        how="left",
        on=["champion", "role"],
        suffixes=("", "_wr")
    )

    # Scoring vectoriel
    server_score = np.where(df_merged["server_wr"] == df_merged["server"], 3,
                    np.where(df_merged["server_wr"] == "tout", 1, 0))

    elo_score = np.where(df_merged["elo_wr"] == df_merged["elo"], 3,
                 np.where(df_merged["elo_wr"] == "tout", 1, 0))

    patch_score = np.zeros(len(df_merged))

    patch = df_merged["patch_float"]
    patch_wr = df_merged["patch_float_wr"]

    # exact
    patch_score[patch == patch_wr] = 4

    # même 15.17
    patch_minor = patch.astype(str).str[:4]
    patch_wr_minor = patch_wr.astype(str).str[:4]
    mask_minor = (patch_score == 0) & (patch_minor == patch_wr_minor)
    patch_score[mask_minor] = 3

    # même 15
    patch_major = patch.astype(str).str[:2]
    patch_wr_major = patch_wr.astype(str).str[:2]
    mask_major = (patch_score == 0) & (patch_major == patch_wr_major)
    patch_score[mask_major] = 2

    # patch restant
    patch_score[patch_score == 0] = 1

    df_merged["score"] = server_score + elo_score + patch_score
    df_merged["patch_distance"] = (patch - patch_wr).abs()

    return df_merged
```

```python
def select_best_per_role(df_scored):
    # trouver l'index du meilleur score par match_id + role
    idx = df_scored.groupby(["match_id","role"]) \
                   .apply(lambda x: x.sort_values(["score","patch_distance"], ascending=[False,True]).index[0])

    df_best = df_scored.loc[idx.values].reset_index(drop=True)

    return df_best
```

```python
def build_df_enhanced(df_result_norm, df_simple_wr_norm):
    # 1️⃣ index vectorisé
    df_indexed = build_champion_index_vector(df_simple_wr_norm)

    # 2️⃣ stack champions
    df_stacked = stack_champions(df_result_norm)

    # 3️⃣ scoring vectoriel
    df_scored = score_candidates(df_stacked, df_indexed)

    # 4️⃣ select best
    df_best = select_best_per_role(df_scored)

    # 5️⃣ pivot pour retrouver les colonnes par role
    extra_cols = [c for c in df_best.columns if c not in ["match_id","role","champion",
                                                          "server","patch","elo","blue_side_win"]]

    df_pivot = df_best.pivot(index="match_id", columns="role", values=extra_cols)
    # flatten multiindex
    df_pivot.columns = [f"{role}_{col}" for col, role in df_pivot.columns]

    # 6️⃣ concat avec df_result_norm
    df_enhanced = df_result_norm.set_index("match_id").join(df_pivot).reset_index()

    return df_enhanced
```

```python
# ICI C'est la bonne methode optimisée en temps mais pas en memoire donc à executer par parties. 
# et peut-etre faudrait-il empecher de ramener bronz en wr si le elo de la game est trop élevé.
import pandas as pd

def build_champion_index_vector(df_simple_wr_norm):
    """
    Prépare le dataframe WR avec patch_float
    et index champion + role
    """
    df = df_simple_wr_norm.copy()

    df["patch_float"] = df["patch"].astype(float)

    # l'index permet de faire la jointure plus rapidement mais le nombre de lignes pendant la jointure explose quand meme.
    # essayer join plutot que merge mais il faut d'abord setter en tant qu'index role et champion dnas les 2 df car join marche sur les index

    df.set_index(["champion","role"], inplace=True)

    return df

def stack_champions(df_result_norm):

    champ_cols = [
        "blue_top_champion","blue_jun_champion","blue_mid_champion",
        "blue_adc_champion","blue_sup_champion",
        "red_top_champion","red_jun_champion","red_mid_champion",
        "red_adc_champion","red_sup_champion"
    ]

    df_stacked = df_result_norm.melt(
        id_vars=["match_id","server","patch","elo","blue_side_win"],
        value_vars=champ_cols,
        var_name="champion_col",
        value_name="champion"
    )

    # extraire team et role
    split = df_stacked["champion_col"].str.split("_", expand=True)

    df_stacked["team"] = split[0]
    df_stacked["role"] = split[1]

    df_stacked["patch_float"] = df_stacked["patch"].astype(float)

    return df_stacked


def score_candidates(df_stacked, df_simple_wr_norm_indexed):

    import numpy as np

    df_wr = df_simple_wr_norm_indexed.reset_index()

    df_merged = df_stacked.merge(
        df_wr,
        how="left",
        on=["champion","role"],
        suffixes=("", "_wr")
    )

    # server score
    server_score = np.where(
        df_merged["server_wr"] == df_merged["server"], 3,
        np.where(df_merged["server_wr"] == "tout", 1, 0)
    )

    # elo score
    elo_score = np.where(
        df_merged["elo_wr"] == df_merged["elo"], 3,
        np.where(df_merged["elo_wr"] == "tout", 1, 0)
    )

    # patch score
    patch = df_merged["patch_float"]
    patch_wr = df_merged["patch_float_wr"]

    patch_score = np.zeros(len(df_merged))

    # exact
    patch_score[patch == patch_wr] = 4

    # même mineur
    patch_minor = patch.astype(str).str[:4]
    patch_wr_minor = patch_wr.astype(str).str[:4]

    mask_minor = (patch_score == 0) & (patch_minor == patch_wr_minor)
    patch_score[mask_minor] = 3

    # même majeur
    patch_major = patch.astype(str).str[:2]
    patch_wr_major = patch_wr.astype(str).str[:2]

    mask_major = (patch_score == 0) & (patch_major == patch_wr_major)
    patch_score[mask_major] = 2

    patch_score[patch_score == 0] = 1

    df_merged["score"] = server_score + elo_score + patch_score
    df_merged["patch_distance"] = (patch - patch_wr).abs()

    return df_merged


def select_best_per_role(df_scored):

    idx = df_scored.groupby(["match_id","team","role"]) \
        .apply(lambda x: x.sort_values(
            ["score","patch_distance"],
            ascending=[False,True]
        ).index[0])

    df_best = df_scored.loc[idx.values].reset_index(drop=True)

    return df_best


def build_df_enhanced(df_result_norm, df_simple_wr_norm):

    # 1 index WR
    df_indexed = build_champion_index_vector(df_simple_wr_norm)

    # 2 stack
    df_stacked = stack_champions(df_result_norm)

    # 3 scoring
    df_scored = score_candidates(df_stacked, df_indexed)

    # 4 best candidate
    df_best = select_best_per_role(df_scored)

    # colonnes à ajouter
    extra_cols = [
        c for c in df_best.columns
        # if c not in [
        #     "match_id","champion","team","role",
        #     "server","patch","elo","blue_side_win",
        #     "champion_col"
        # ]
        if c not in [
            "match_id",
            "champion",
            "team",
            "role",
            "server","patch","elo","blue_side_win",
            "champion_col"
        ]
    ]

    # clé pivot
    df_best["team_role"] = df_best["team"] + "_" + df_best["role"]

    df_pivot = df_best.pivot(
        index="match_id",
        columns="team_role",
        values=extra_cols
    )

    # flatten colonnes
    df_pivot.columns = [
        f"{team_role}_{col}"
        for col, team_role in df_pivot.columns
    ]

    df_enhanced = (
        df_result_norm
        .set_index("match_id")
        .join(df_pivot)
        .reset_index()
    )

    return df_enhanced
```

```python
################### optimisé en memoire
def build_df_enhanced_low_mem(df_result_norm, df_simple_wr_norm):
    import pandas as pd
    import numpy as np

    champion_index = {champ: g for champ, g in df_simple_wr_norm.groupby("champion")}

    df_result_norm = df_result_norm.copy()
    roles = [
        "blue_top", "blue_jun", "blue_mid", "blue_adc", "blue_sup",
        "red_top", "red_jun", "red_mid", "red_adc", "red_sup"
    ]

    for role in roles:

        champ_col = f"{role}_champion"
        champions = df_result_norm[champ_col].values

        merged_rows = []

        for i, champ in enumerate(champions):

            if champ not in champion_index:
                merged_rows.append({})
                continue

            df_cand = champion_index[champ].copy()
            row = df_result_norm.iloc[i]

            # Vectorial scoring sur ce petit df
            server_score = np.where(df_cand["server"] == row["server"], 3,
                            np.where(df_cand["server"] == "tout", 1, 0))
            elo_score = np.where(df_cand["elo"] == row["elo"], 3,
                         np.where(df_cand["elo"] == "tout", 1, 0))
            patch_score = np.zeros(len(df_cand))
            patch_float = float(row["patch"])
            patch_cand_float = df_cand["patch"].astype(float)
            patch_score[patch_cand_float == patch_float] = 4
            patch_minor = ".".join(str(row["patch"]).split(".")[:2])
            mask_minor = df_cand["patch"].astype(str).str.startswith(patch_minor)
            patch_score[(patch_score == 0) & mask_minor] = 3
            patch_major = str(row["patch"]).split(".")[0]
            mask_major = df_cand["patch"].astype(str).str.startswith(patch_major)
            patch_score[(patch_score == 0) & mask_major] = 2
            patch_score[patch_score == 0] = 1

            total_score = server_score + elo_score + patch_score
            best_idx = np.argmax(total_score)
            best_row = df_cand.iloc[best_idx]

            # ajouter les colonnes avec prefix role
            new_data = {f"{role}_{c}": best_row[c] for c in df_cand.columns if c != "champion"}
            merged_rows.append(new_data)

        df_extra = pd.DataFrame(merged_rows)
        df_result_norm = pd.concat([df_result_norm.reset_index(drop=True), df_extra], axis=1)

    return df_result_norm
```

```python
def add_blue_top_features(df_result_norm, df_simple_wr_norm):
    import pandas as pd

    # --- sécuriser les types ---
    for col in ["server", "patch", "elo"]:
        df_result_norm[col] = df_result_norm[col].astype(str)
        df_simple_wr_norm[col] = df_simple_wr_norm[col].astype(str)

    df_result_norm["blue_top_champion"] = df_result_norm["blue_top_champion"].astype(str)
    df_simple_wr_norm["champion"] = df_simple_wr_norm["champion"].astype(str)

    # --- merge vectorisé ---
    df_merged = df_result_norm.merge(
        df_simple_wr_norm,
        how="left",
        left_on=["server", "patch", "elo", "blue_top_champion"],
        right_on=["server", "patch", "elo", "champion"],
        suffixes=("", "_simple")
    )

    # --- debug ---
    exact_mask = df_merged["champion"].notna()

    for _, row in df_merged[exact_mask].iterrows():
        print("exact match :", row["server"], row["patch"], row["elo"], row["blue_top_champion"])

    for _, row in df_merged[~exact_mask].iterrows():
        print("pas d'exact match :", row["server"], row["patch"], row["elo"], row["blue_top_champion"])

    # --- colonnes à ajouter ---
    cols_to_add = [
        c for c in df_simple_wr_norm.columns
        if c not in ["server", "patch", "elo", "champion"]
    ]

    rename_dict = {c: f"blue_top_{c}" for c in cols_to_add}

    df_merged = df_merged.rename(columns=rename_dict)

    new_cols = list(rename_dict.values())

    df_final = pd.concat(
        [df_result_norm.reset_index(drop=True),
         df_merged[new_cols].reset_index(drop=True)],
        axis=1
    )

    return df_final
```
