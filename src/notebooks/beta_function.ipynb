```python
import pandas as pd
import numpy as np


def normalize_str(x):
    if pd.isna(x):
        return None
    return str(x).strip().lower().replace(".", "")


def complete_role_play_ratio(matchups_df, df_simple_wr, target_patch):
    
    df_simple_wr_copy = df_simple_wr.copy()
    
    # normalisation pour matching robuste
    df_simple_wr_copy["name_norm"] = df_simple_wr_copy["name"].apply(normalize_str)
    df_simple_wr_copy["role_norm"] = df_simple_wr_copy["role"].apply(normalize_str)
    
    matchups_df["champion_norm"] = matchups_df["champion"].apply(normalize_str)
    matchups_df["role_norm"] = matchups_df["role"].apply(normalize_str)
    
    # lignes à corriger
    mask_missing = matchups_df["role_play_ratio"] == "-"
    rows_to_fix = matchups_df[mask_missing]
    
    print(f"{len(rows_to_fix)} lignes à corriger\n")
    print(rows_to_fix)
    
    for idx, row in rows_to_fix.iterrows():
        
        champ = row["champion"]
        role = row["role"]
        
        champ_norm = row["champion_norm"]
        role_norm = row["role_norm"]
        
        print(f"\n--- Traitement : {champ} ({role}) ---")
        
        # filtrage strict
        candidates = df_simple_wr_copy[
            (df_simple_wr_copy["server"] == "TOUT") &
            (df_simple_wr_copy["elo"] == "TOUT") &
            # (df_simple_wr_copy["patch"] == target_patch) &
            (df_simple_wr_copy["patch"].astype(str).str.strip() == str(target_patch).strip()) &
            (df_simple_wr_copy["name_norm"] == champ_norm) &
            (df_simple_wr_copy["role_norm"] == role_norm)
        ]
        
        # aucun candidat exact
        if len(candidates) == 0:
            print("❌ Aucun match exact trouvé.")
            
            # fallback : ignorer patch
            candidates_loose = df_simple_wr_copy[
                (df_simple_wr_copy["name_norm"] == champ_norm) &
                (df_simple_wr_copy["role_norm"] == role_norm)
            ]
            
            if len(candidates_loose) > 0:
                print("⚠️ Candidats proches trouvés (sans filtre patch/server/elo):")
                print(candidates_loose[["server","elo","patch","name","role","role_pickrate"]])
            else:
                print("🚨 Aucun candidat trouvé du tout.")
            
            continue
        
        # plusieurs candidats
        if len(candidates) > 1:
            print("⚠️ Plusieurs candidats exacts trouvés :")
            print(candidates[["server","elo","patch","name","role","role_pickrate"]])
        
        # on prend le premier
        selected = candidates.iloc[0]
        
        print(f"✅ Match trouvé → role_pickrate = {selected['role_pickrate']}")
        
        # remplacement
        matchups_df.at[idx, "role_play_ratio"] = selected["role_pickrate"]
    
    # nettoyage colonnes temporaires
    matchups_df.drop(columns=["champion_norm","role_norm"], inplace=True)
    
    return matchups_df
```

```python
# import pandas as pd

# def explode_matchups(df, lane_prefix):
#     rows = []
#     cols_to_keep = [
#         'role_play_ratio', 'tier', 'rank', 'winrate', 
#         'pickrate', 'banrate', 'nb_games_analyzed'
#     ]
    
#     for _, row in df.iterrows():
#         champ = row["champion"]
#         role = row["role"]
        
#         # Récupération des colonnes à conserver
#         kept_data = {col: row[col] for col in cols_to_keep if col in df.columns}
        
#         for i in range(1, 5):
#             name_col = f"{lane_prefix}_{i}_name"
#             wr_col = f"{lane_prefix}_{i}_winrate"
#             games_col = f"{lane_prefix}_{i}_games"
#             relative_role_col = lane_prefix
            
#             if name_col in df.columns and pd.notna(row[name_col]):
#                 new_row = {
#                     "champion": champ,
#                     "role": role,
#                     f"{lane_prefix.split('_')[0]}": row[name_col],
#                     # "opponent": row[name_col],
#                     "winrate_relatif": float(str(row[wr_col]).replace("%","")) if pd.notna(row[wr_col]) else None,
#                     "relative_role": relative_role_col.split("_")[1],  # extrait "adc", "top", etc. de "matchup_adc"
#                     "games": row[games_col] if pd.notna(row[games_col]) else None
#                 }
#                 # Ajouter les colonnes à conserver
#                 new_row.update(kept_data)
                
#                 rows.append(new_row)
                
#     return pd.DataFrame(rows)


import pandas as pd
import re

def explode_matchups(df, lane_prefix):
    rows = []
    
    cols_to_keep = [
        'role_play_ratio', 'tier', 'rank', 'winrate', 
        'pickrate', 'banrate', 'nb_games_analyzed'
    ]
    
    # 🔎 Trouver dynamiquement tous les index existants
    pattern = re.compile(f"{lane_prefix}_(\\d+)_name")
    
    matchup_indices = sorted([
        int(pattern.search(col).group(1))
        for col in df.columns
        if pattern.search(col)
    ])
    
    for _, row in df.iterrows():
        
        champ = row["champion"]
        role = row["role"]
        
        kept_data = {col: row[col] for col in cols_to_keep if col in df.columns}
        
        for i in matchup_indices:
            
            name_col = f"{lane_prefix}_{i}_name"
            wr_col = f"{lane_prefix}_{i}_winrate"
            games_col = f"{lane_prefix}_{i}_games"
            
            if name_col in df.columns and pd.notna(row[name_col]):
                
                winrate_value = row[wr_col]
                
                if pd.notna(winrate_value):
                    winrate_value = float(str(winrate_value).replace("%", ""))
                else:
                    winrate_value = None
                
                new_row = {
                    "champion": champ,
                    "role": role,
                    "matchup": row[name_col],
                    "winrate_relatif": winrate_value,
                    "relative_role": lane_prefix.split("_")[1],
                    "games": row[games_col] if pd.notna(row[games_col]) else None
                }
                
                new_row.update(kept_data)
                rows.append(new_row)
    
    return pd.DataFrame(rows)
```

```python
def get_alpha_beta(row):

    opponent_col = row.index[2]  # récupère le nom de la 3ème colonne
    opponent_value = row[opponent_col]
    print(f"Calculating alpha/beta for {row['champion']} ({row['role']}) vs ({opponent_col}) {opponent_value} ({row['relative_role']})")
    print(f"Winrate relatif: {row['winrate_relatif']}, Games: {row['games']}")
    winrate = row["winrate_relatif"] / 100
    games = row["games"]
    
    if pd.isna(winrate) or pd.isna(games) or games == 0:
        return pd.Series({"alpha": None, "beta": None})
    
    # Convertir le winrate en nombre de victoires et défaites
    victories = winrate * games
    defeats = (1 - winrate) * games
    
    # Ajouter un petit epsilon pour éviter les problèmes de division par zéro
    epsilon = 1e-6
    alpha = victories + epsilon
    beta = defeats + epsilon
    
    return pd.Series({"alpha": alpha, "beta": beta})
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# def plot_beta(alpha, beta_param, title=None, color='blue'):
#     """
#     Trace une distribution Beta pour les paramètres alpha et beta donnés.
    
#     Args:
#         alpha (float): paramètre alpha de la Beta
#         beta_param (float): paramètre beta de la Beta
#         title (str, optional): titre du graphique
#         color (str, optional): couleur de la courbe
#     """
#     x = np.linspace(0, 1, 1000)
#     y = beta.pdf(x, alpha, beta_param)
    
#     # Normalisation pour que l'aire = 1
#     y /= np.trapz(y, x)
    
#     plt.figure(figsize=(10, 6))
#     plt.plot(x, y, color=color, label=f'Beta(alpha={alpha:.2f}, beta={beta_param:.2f})')
#     plt.fill_between(x, 0, y, color=color, alpha=0.2)
#     plt.xlabel('Winrate')
#     plt.ylabel('Densité de probabilité')
#     plt.title(title if title else f'Distribution Beta (alpha={alpha:.2f}, beta={beta_param:.2f})')
#     plt.grid(True)
#     plt.legend()
#     plt.show()



def plot_beta(alpha_beta_param, title=None, color='blue'):
    """
    Trace une distribution Beta pour les paramètres alpha et beta donnés.
    
    Args:
        alpha (float): paramètre alpha de la Beta
        beta_param (float): paramètre beta de la Beta
        title (str, optional): titre du graphique
        color (str, optional): couleur de la courbe
    """
    x = np.linspace(0, 1, 1000)
    y = beta.pdf(x, alpha_beta_param[0], alpha_beta_param[1])
    
    # Normalisation pour que l'aire = 1
    y /= np.trapz(y, x)
    
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, color=color, label=f'Beta(alpha={alpha_beta_param[0]:.2f}, beta={alpha_beta_param[1]:.2f})')
    plt.fill_between(x, 0, y, color=color, alpha=0.2)
    plt.xlabel('Winrate')
    plt.ylabel('Densité de probabilité')
    plt.title(title if title else f'Distribution Beta (alpha={alpha_beta_param[0]:.2f}, beta={alpha_beta_param[1]:.2f})')
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

def combine_beta_distributions(alpha1, beta1, alpha2, beta2, corr=-1, n_points=1000):
    """
    Combine deux distributions Beta en une distribution résultante selon le paramètre corr.
    
    corr:
        -1   => indépendantes
         0   => totalement indépendantes (équivalent à -1)
         1   => peu corrélées (~1%)
        10   => corrélation 10%
        50   => corrélation 50%
        75   => corrélation 75%
        100  => totalement corrélées (identiques)
    """
    x = np.linspace(0, 1, n_points)
    
    # Pdf des distributions initiales
    pdf1 = beta.pdf(x, alpha1, beta1)
    pdf2 = beta.pdf(x, alpha2, beta2)
    
    # Normalisation
    pdf1 /= np.trapz(pdf1, x)
    pdf2 /= np.trapz(pdf2, x)
    
    # Combinaison selon corr
    if corr == -1 or corr == 0:
        # Indépendantes : on multiplie les densités (et renormalise)
        combined_pdf = pdf1 * pdf2
        combined_pdf /= np.trapz(combined_pdf, x)
    elif corr >= 100:
        # Totalement corrélées : la distribution est la même
        combined_pdf = pdf1
    else:
        # Corrélation partielle : moyenne pondérée
        weight = min(max(corr/100, 0), 1)  # corr=50 => poids 0.5
        combined_pdf = (1 - weight) * pdf1 + weight * pdf2
        combined_pdf /= np.trapz(combined_pdf, x)  # normalisation
    
    return x, combined_pdf

def plot_combined_beta(alpha1, beta1, alpha2, beta2, corr=-1):
    x, combined_pdf = combine_beta_distributions(alpha1, beta1, alpha2, beta2, corr)
    
    plt.figure(figsize=(10,6))
    plt.plot(x, combined_pdf, label=f'Combined Beta (corr={corr})', color='green')
    plt.xlabel('Winrate')
    plt.ylabel('Densité de probabilité')
    plt.title('Distribution Beta combinée')
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
# Beta exemple
alpha1, beta1 = 50, 30       # matchup spécifique
alpha2, beta2 = 30, 50       # winrate global

# Combinaison indépendante
plot_combined_beta(alpha1, beta1, alpha2, beta2, corr=1)

# Combinaison corrélation 50%
plot_combined_beta(alpha1, beta1, alpha2, beta2, corr=50)

# Combinaison totalement corrélée
plot_combined_beta(alpha1, beta1, alpha2, beta2, corr=75)
```

```python
import numpy as np
from scipy.stats import beta
from scipy.stats import skew, kurtosis

def compute_combined_betas_from_list_sym(
    alpha_beta_list,
    corr=50,
    n_points=100000
):
    """
    Combine plusieurs distributions Beta et retourne :
    - x
    - combined_pdf
    - statistiques interprétables
    """

    x = np.linspace(0, 1, n_points)
    pdfs = []
    epsilon = 1e-6

    # PDFs individuelles
    for alpha, beta_param in alpha_beta_list:
        alpha_safe = max(alpha, epsilon)
        beta_safe = max(beta_param, epsilon)
        pdf = beta.pdf(x, alpha_safe, beta_safe)
        pdf /= np.trapz(pdf, x)
        pdfs.append(pdf)

    # Combinaison symétrique
    combined_pdf = sum(pdfs) / len(pdfs)
    combined_pdf /= np.trapz(combined_pdf, x)

    # ===== ANALYSE =====

    dx = x[1] - x[0]

    mean = np.trapz(x * combined_pdf, x)
    variance = np.trapz((x - mean)**2 * combined_pdf, x)
    std = np.sqrt(variance)

    cdf = np.cumsum(combined_pdf) * dx

    median = x[np.searchsorted(cdf, 0.5)]
    p05 = x[np.searchsorted(cdf, 0.05)]
    p95 = x[np.searchsorted(cdf, 0.95)]
    p25 = x[np.searchsorted(cdf, 0.25)]
    p75 = x[np.searchsorted(cdf, 0.75)]

    mode = x[np.argmax(combined_pdf)]

    skewness = skew(combined_pdf)
    kurt = kurtosis(combined_pdf)

    prob_above_50 = np.trapz(combined_pdf[x >= 0.5], x[x >= 0.5])
    prob_above_60 = np.trapz(combined_pdf[x >= 0.6], x[x >= 0.6])

    entropy = -np.trapz(
        combined_pdf * np.log(combined_pdf + 1e-12),
        x
    )

    stats = {
        "mean": mean,
        "median": median,
        "mode": mode,
        "std": std,
        "variance": variance,
        "skewness": skewness,
        "kurtosis": kurt,
        "entropy": entropy,
        "credible_interval_5_95": (p05, p95),
        "iqr_25_75": (p25, p75),
        "prob_winrate_above_50": prob_above_50,
        "prob_winrate_above_60": prob_above_60,
    }

    return x, combined_pdf, stats
```

```python
def plot_combined_betas_from_list_sym(alpha_beta_list, corr=50, n_points=1000):
    """
    Combine plusieurs distributions Beta données sous forme de liste de tuples (alpha, beta)
    et trace la distribution combinée sur le graphique.
    Cette version est **symétrique**, le résultat ne dépend pas de l'ordre.
    
    Args:
        alpha_beta_list (list of tuples): [(alpha1, beta1), (alpha2, beta2), ...]
        corr (float): corrélation partielle en % (default=50)
        n_points (int): nombre de points pour le tracé
    """
    x = np.linspace(0, 1, n_points)
    pdfs = []
    epsilon = 1e-6  # pour éviter alpha=0 ou beta=0
    
    # Calcul des PDFs individuelles
    for alpha, beta_param in alpha_beta_list:
        alpha_safe = max(alpha, epsilon)
        beta_safe = max(beta_param, epsilon)
        pdf = beta.pdf(x, alpha_safe, beta_safe)
        pdf /= np.trapz(pdf, x)  # normalisation
        pdfs.append(pdf)
    
    # Combinaison symétrique : moyenne pondérée de toutes les PDFs
    weight = min(max(corr/100, 0), 1)
    combined_pdf = sum(pdfs) / len(pdfs)  # moyenne simple
    # appliquer la corrélation partielle si corr != 100
    if corr < 100:
        # on renforce la moyenne pondérée vers chaque PDF individuellement
        combined_pdf = (1 - weight) * combined_pdf + weight * sum(pdfs)/len(pdfs)
    
    combined_pdf /= np.trapz(combined_pdf, x)  # normalisation finale
    
    # Tracé
    plt.figure(figsize=(10,6))
    for i, pdf in enumerate(pdfs):
        plt.plot(x, pdf, linestyle='--', alpha=0.5, label=f'Beta {i+1}')
    
    plt.plot(x, combined_pdf, color='green', label=f'Combined Beta (corr={corr}%)')
    plt.fill_between(x, 0, combined_pdf, color='green', alpha=0.2)
    
    plt.xlabel('Winrate')
    plt.ylabel('Densité de probabilité')
    plt.title('Distribution Beta combinée (symétrique)')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
def extract_matchups_and_synergies(row):
    
    # --- 1️⃣ Construction équipes avec rôle ---
    
    blue_team = [
        ("blue", "top", row["blue_top_champion"]),
        ("blue", "jun", row["blue_jungle_champion"]),
        ("blue", "mid", row["blue_mid_champion"]),
        ("blue", "adc", row["blue_adc_champion"]),
        ("blue", "sup", row["blue_support_champion"]),
    ]
    
    red_team = [
        ("red", "top", row["red_top_champion"]),
        ("red", "jun", row["red_jungle_champion"]),
        ("red", "mid", row["red_mid_champion"]),
        ("red", "adc", row["red_adc_champion"]),
        ("red", "sup", row["red_support_champion"]),
    ]
    
    relations = []
    
    # --- 2️⃣ VS (matchups inter-équipes) ---
    
    # Blue vs Red
    for b_side, b_role, b_champ in blue_team:
        for r_side, r_role, r_champ in red_team:
            
            left = f"{b_side}_{b_role}_{b_champ}"
            right = f"{r_side}_{r_role}_{r_champ}"
            
            relations.append(f"{left}_vs_{right}")
    
    # Red vs Blue
    for r_side, r_role, r_champ in red_team:
        for b_side, b_role, b_champ in blue_team:
            
            left = f"{r_side}_{r_role}_{r_champ}"
            right = f"{b_side}_{b_role}_{b_champ}"
            
            relations.append(f"{left}_vs_{right}")
    
    
    # --- 3️⃣ FEAT (synergies intra-équipe) ---
    
    # Blue synergies
    for i in range(len(blue_team)):
        for j in range(len(blue_team)):
            if i != j:
                side_i, role_i, champ_i = blue_team[i]
                side_j, role_j, champ_j = blue_team[j]
                
                left = f"{side_i}_{role_i}_{champ_i}"
                right = f"{side_j}_{role_j}_{champ_j}"
                
                relations.append(f"{left}_feat_{right}")
    
    # Red synergies
    for i in range(len(red_team)):
        for j in range(len(red_team)):
            if i != j:
                side_i, role_i, champ_i = red_team[i]
                side_j, role_j, champ_j = red_team[j]
                
                left = f"{side_i}_{role_i}_{champ_i}"
                right = f"{side_j}_{role_j}_{champ_j}"
                
                relations.append(f"{left}_feat_{right}")
    
    
    assert len(relations) == 90, f"{len(relations)} relations générées au lieu de 90"
    
    return relations
```

```python
def get_winrates_from_relations_debug_v2(
    relations,
    adc_matchups, mid_matchups, top_matchups, jun_matchups, sup_matchups,
    adc_synergy, mid_synergy, top_synergy, jun_synergy, sup_synergy
):
    
    matchup_dfs = {
        "adc": adc_matchups,
        "mid": mid_matchups,
        "top": top_matchups,
        "jun": jun_matchups,
        "sup": sup_matchups,
    }
    
    synergy_dfs = {
        "adc": adc_synergy,
        "mid": mid_synergy,
        "top": top_synergy,
        "jun": jun_synergy,
        "sup": sup_synergy,
    }
    
    winrates = []
    nombre_games = []
    
    for idx, rel in enumerate(relations):
        print(f"\n--- Relation {idx}: {rel} ---")
        
        try:
            if "_vs_" in rel:
                left, right = rel.split("_vs_")
                df_dict = matchup_dfs
                col_match = "matchup"
            
            elif "_feat_" in rel:
                left, right = rel.split("_feat_")
                df_dict = synergy_dfs
                col_match = "matchup"
            
            else:
                print("Relation non reconnue → 0.5")
                winrates.append(0.5)
                nombre_games.append(2)
                continue
            
            # 🔹 Parse
            side_l, role_l, champ_l = left.split("_", 2)
            side_r, role_r, champ_r = right.split("_", 2)
            
            role_l = role_l[:3]
            role_r = role_r[:3]
            
            df = df_dict.get(role_r)
            
            if df is None:
                print(f"Aucun DF pour rôle {role_r} → 0.5")
                winrates.append(0.5)
                nombre_games.append(2)
                continue
            
            result = df[
                (df["champion"] == champ_l) &
                (df["role"] == role_l) &
                (df[col_match] == champ_r) &
                (df["relative_role"] == role_r)
            ]
            
            if len(result) >= 1:
                wr = result.iloc[0]["winrate_relatif"] / 100.0
                nb_games = result.iloc[0]["games"]
                
                # 🔥 LOGIQUE D'INVERSION
                if side_l == "red":
                    wr = 1 - wr
                    print(f"Inversion car red en premier → {wr}")
                
                winrates.append(wr)
                nombre_games.append(nb_games)
            
            else:
                winrates.append(0.5)
                nombre_games.append(2)
        
        except Exception as e:
            print(f"Erreur: {e} → 0.5")
            winrates.append(0.5)
            nombre_games.append(2)
    
    return winrates, nombre_games
```

```python
import pandas as pd
import numpy as np

def add_alpha_beta_columns(df, epsilon=1e-6, debug=True):
    
    print("===== START add_alpha_beta_columns =====")
    
    df_copy = df.copy()
    df_copy = df_copy.iloc[:, [0]]
    
    # Toutes les colonnes winrate
    wr_cols = [c for c in df.columns if c.startswith("wr_")]
    
    print(f"Nombre de colonnes wr_ détectées : {len(wr_cols)}")
    
    if debug:
        print("Exemple wr_cols :", wr_cols[:5])
    
    for wr_col in wr_cols:
        
        games_col = wr_col.replace("wr_", "nb_games_")
        
        if games_col not in df.columns:
            print(f"⚠️ Colonne games manquante pour {wr_col} → attendu: {games_col}")
            continue
        
        alpha_col = wr_col.replace("wr_", "alpha_")
        beta_col = wr_col.replace("wr_", "beta_")
        
        if debug:
            print(f"\n--- Traitement {wr_col} ---")
            print(f"Games column : {games_col}")
            print(f"Alpha column : {alpha_col}")
            print(f"Beta column  : {beta_col}")
        
        winrate = df[wr_col]
        games = df[games_col]
        
        if debug:
            print("Type winrate :", winrate.dtype)
            print("Type games   :", games.dtype)
            print("NaN winrate :", winrate.isna().sum())
            print("NaN games   :", games.isna().sum())
        
        # Conversion safe en float
        winrate = pd.to_numeric(winrate, errors="coerce")
        games = pd.to_numeric(games, errors="coerce")
        
        victories = winrate * games
        defeats = (1 - winrate) * games
        
        if debug:
            print("Exemple victories :", victories.head(2).tolist())
            print("Exemple defeats   :", defeats.head(2).tolist())
        
        df_copy[alpha_col] = victories + epsilon
        df_copy[beta_col] = defeats + epsilon
    
    print("===== END add_alpha_beta_columns =====")
    
    return df_copy
```

```python
import re

def extract_all_alpha_beta_lists_from_row(row, debug=False):
    """
    À partir d'une ligne contenant les colonnes alpha_* et beta_*,
    retourne :
    
    - 10 listes (1 par champion) contenant 9 couples (alpha, beta)
    - 1 liste globale contenant les 90 couples
    """

    # 🔹 récupérer toutes les colonnes alpha
    alpha_cols = [c for c in row.index if c.startswith("alpha_")]

    if debug:
        print(f"Nombre total colonnes alpha : {len(alpha_cols)}")

    # 🔹 identifier les 10 champions uniques
    # format colonne = alpha_blue_top_vs_red_mid
    champion_prefixes = set()

    for col in alpha_cols:
        match = re.match(r"alpha_(blue|red)_(top|jun|mid|adc|sup)", col)
        if match:
            champion_prefixes.add(f"{match.group(1)}_{match.group(2)}")

    champion_prefixes = sorted(list(champion_prefixes))

    if debug:
        print("Champions détectés :", champion_prefixes)
        print("Nombre champions :", len(champion_prefixes))

    champion_lists = []
    global_list = []

    # 🔹 pour chaque champion
    for champ in champion_prefixes:

        champ_alpha_cols = sorted([
            c for c in alpha_cols if c.startswith(f"alpha_{champ}")
        ])

        champ_list = []

        for alpha_col in champ_alpha_cols:

            beta_col = alpha_col.replace("alpha_", "beta_")

            if beta_col not in row.index:
                continue

            alpha = row[alpha_col]
            beta = row[beta_col]

            champ_list.append((alpha, beta))
            global_list.append((alpha, beta))

        if debug:
            print(f"{champ} → {len(champ_list)} couples")

        champion_lists.append(champ_list)

    if debug:
        print("Total global couples :", len(global_list))

    # retourne 10 listes + 1 globale
    return champion_lists + [global_list]
```

```python
def plot_combined_betas_from_list_sym_better(alpha_beta_list, corr=50, n_points=1000):
    """
    Combine plusieurs distributions Beta données sous forme de liste de tuples (alpha, beta)
    et trace la distribution combinée sur le graphique.
    Version zoomée sur [0.33, 0.66] avec repères visuels.
    """
    import numpy as np
    import matplotlib.pyplot as plt
    from scipy.stats import beta
    
    x = np.linspace(0, 1, n_points)
    pdfs = []
    epsilon = 1e-6
    
    # 🔹 Calcul des PDFs individuelles
    for alpha, beta_param in alpha_beta_list:
        alpha_safe = max(alpha, epsilon)
        beta_safe = max(beta_param, epsilon)
        print(alpha_safe, beta_safe)
        pdf = beta.pdf(x, alpha_safe, beta_safe)
        print(pdf)
        pdf /= np.trapz(pdf, x)
        print(pdf)
        pdfs.append(pdf)
    
    # 🔹 Combinaison symétrique
    weight = min(max(corr/100, 0), 1)
    combined_pdf = sum(pdfs) / len(pdfs)
    
    if corr < 100:
        combined_pdf = (1 - weight) * combined_pdf + weight * (sum(pdfs) / len(pdfs))
    
    combined_pdf /= np.trapz(combined_pdf, x)
    
    # 🔹 Plot
    plt.figure(figsize=(10,6))
    
    for i, pdf in enumerate(pdfs):
        plt.plot(x, pdf, linestyle='--', alpha=0.4)
    
    plt.plot(x, combined_pdf, linewidth=2)
    plt.fill_between(x, 0, combined_pdf, alpha=0.2)
    
    # 🔥 ZOOM
    plt.xlim(0.40, 0.60)
    
    # 🔥 Repères verticaux
    markers = [0.45, 0.50, 0.55]
    for m in markers:
        plt.axvline(x=m, linestyle=':', linewidth=1.5)
        plt.text(m, 0, f"{m:.2f}", 
                 horizontalalignment='center',
                 verticalalignment='bottom')
    
    plt.xlabel("Winrate")
    plt.ylabel("Densité de probabilité")
    plt.title("Distribution Beta combinée (zoom 0.40 - 0.60)")
    plt.grid(True)
    
    plt.show()
```
