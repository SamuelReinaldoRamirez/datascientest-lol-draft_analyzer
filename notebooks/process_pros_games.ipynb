```python
def get_puuid(game_name, tag_line, region):
    url = f"https://{region}.api.riotgames.com/riot/account/v1/accounts/by-riot-id/{game_name}/{tag_line}"
    r = requests.get(url, headers=HEADERS)
    r.raise_for_status()
    return r.json()["puuid"]
```

```python
def get_last_matches(puuid, region="europe", count=10):
    url = f"https://{region}.api.riotgames.com/lol/match/v5/matches/by-puuid/{puuid}/ids"
    params = {"start": 0, "count": count}
    r = requests.get(url, headers=HEADERS, params=params)
    r.raise_for_status()
    return r.json()
```

```python
import pandas as pd

# Copie de sécurité
df0 = df.copy()

# Colonnes d'identification d'une game
GAME_KEYS = [
    "gameid", "league", "year", "split", "playoffs",
    "date", "game", "patch"
]

# =========================
# 1️⃣ Séparer joueurs / teams
# =========================
df_players = df0[df0["participantid"].between(1, 10)]
df_teams = df0[df0["participantid"].isin([100, 200])]

# =========================
# 2️⃣ Pivot joueurs → 1 ligne par game
# =========================
player_cols = [c for c in df_players.columns if c not in GAME_KEYS + ["participantid"]]

players_wide = (
    df_players
    .set_index(GAME_KEYS + ["participantid"])[player_cols]
    .unstack("participantid")
)

# Aplatir les colonnes (ex: kills_p1, kills_p2, ...)
players_wide.columns = [
    f"{col}_p{pid}" for col, pid in players_wide.columns
]

players_wide = players_wide.reset_index()

# =========================
# 3️⃣ Pivot teams (Blue / Red)
# =========================
team_cols = [c for c in df_teams.columns if c not in GAME_KEYS + ["participantid"]]

teams_wide = (
    df_teams
    .set_index(GAME_KEYS + ["participantid"])[team_cols]
    .unstack("participantid")
)

teams_wide.columns = [
    f"{col}_{'blue' if pid == 100 else 'red'}"
    for col, pid in teams_wide.columns
]

teams_wide = teams_wide.reset_index()

# =========================
# 4️⃣ Fusion finale
# =========================
df_games = (
    players_wide
    .merge(teams_wide, on=GAME_KEYS, how="left")
)

df_games.head()
```
